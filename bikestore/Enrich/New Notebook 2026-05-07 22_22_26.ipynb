```python
%sql

--#################################################################
-- Autor: José Paulo
-- Processo: Carga da tabela Pedidos_OBT
--#################################################################

INSERT INTO bikestore.silver.Pedidos_OBT

SELECT

    ped.order_id AS Nota_Fiscal,

    item.item_id AS Item_Nota_Fiscal,

    prod.product_name AS Nome_Produto,

    CAST(
        ROUND(
            (item.list_price * item.quantity),
            2
        ) AS DOUBLE
    ) AS Valor_Bruto,

    CAST(
        ROUND(
            (
                item.list_price
                * item.quantity
                * item.discount
            ),
            2
        ) AS DOUBLE
    ) AS Valor_Desconto,

    CAST(
        ROUND(
            (
                item.list_price
                * item.quantity
                * (1 - item.discount)
            ),
            2
        ) AS DOUBLE
    ) AS Valor_Liquido,

    CASE
        WHEN ped.order_status = 1 THEN 'Pendente'
        WHEN ped.order_status = 2 THEN 'Em Processamento'
        WHEN ped.order_status = 3 THEN 'Enviado'
        WHEN ped.order_status = 4 THEN 'Entregue'
        ELSE 'Desconhecido'
    END AS Status_Nota_Fiscal,

    ped.order_date AS Data_Pedido,

    ped.required_date AS Data_Entrega,

    ped.shipped_date AS Data_Envio,

    CONCAT(
        staff.first_name,
        ' ',
        staff.last_name
    ) AS Nome_Vendedor,

    staff.email AS Email_Vendedor,

    staff.phone AS Telefone_Vendedor,

    store.store_name AS Nome_Loja,

    store.phone AS Telefone_Loja,

    store.email AS Email_Loja,

    store.street AS Endereco_Loja

FROM bikestore.silver.order_items item

INNER JOIN bikestore.silver.orders ped
    ON ped.order_id = item.order_id

INNER JOIN bikestore.silver.products prod
    ON prod.product_id = item.product_id

INNER JOIN bikestore.silver.stores store
    ON store.store_id = ped.store_id

INNER JOIN bikestore.silver.staffs staff
    ON staff.staff_id = ped.staff_id

;
```
